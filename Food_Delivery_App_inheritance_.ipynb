```python
class Food:
    def __init__(self, name, price, quantity=1):
        self.name = name
        self.price = price
        self.quantity = quantity

    def subtotal(self):
        return self.price * self.quantity

    def preparation_time(self):
        # Base prep time — children override with their own
        return 10

    def discount_rate(self):
        # Base discount — children override
        return 0.0

    def discount(self):
        return self.subtotal() * self.discount_rate()

    def final_price(self):
        return self.subtotal() - self.discount()

    def show_item(self):
        print(f"  {self.name} x{self.quantity} @ Rs. {self.price:.0f} = Rs. {self.subtotal():.0f}")
        if self.discount_rate() > 0:
            print(f"    Discount: Rs. {self.discount():.0f} ({self.discount_rate()*100:.0f}%)")


class Pizza(Food):
    def __init__(self, name="Pizza", price=1200, quantity=1):
        super().__init__(name, price, quantity)

    def preparation_time(self):
        return 25                          # pizzas take longer

    def discount_rate(self):
        return 0.10                        # 10% off pizzas


class Burger(Food):
    def __init__(self, name="Burger", price=500, quantity=1):
        super().__init__(name, price, quantity)

    def preparation_time(self):
        return 15

    def discount_rate(self):
        return 0.05                        # 5% off burgers


class Drink(Food):
    def __init__(self, name="Drink", price=150, quantity=1):
        super().__init__(name, price, quantity)

    def preparation_time(self):
        return 3                           # drinks are fast

    def discount_rate(self):
        return 0.0                         # no discount on drinks


# ---- The order ----
class Order:
    def __init__(self):
        self.items = []                    # list of Food objects

    def add_food(self, food):
        self.items.append(food)
        print(f"Added: {food.name} x{food.quantity}")

    def remove_food(self, index):
        if 0 <= index < len(self.items):
            removed = self.items.pop(index)
            print(f"Removed: {removed.name} x{removed.quantity}")
        else:
            print("Invalid item index.")

    def calculate_bill(self):
        subtotal   = sum(item.subtotal() for item in self.items)
        discount   = sum(item.discount() for item in self.items)
        total      = subtotal - discount
        prep_time  = max((item.preparation_time() for item in self.items), default=0)
        return subtotal, discount, total, prep_time

    def show_order(self):
        print()
        print("===== ORDER SUMMARY =====")
        if not self.items:
            print("  (empty)")
            return

        for i, item in enumerate(self.items, start=1):
            print(f"{i}.", end=" ")
            item.show_item()

        subtotal, discount, total, prep_time = self.calculate_bill()
        print()
        print(f"Subtotal:         Rs. {subtotal:.0f}")
        print(f"Total Discount:   Rs. {discount:.0f}")
        print(f"Final Bill:       Rs. {total:.0f}")
        print(f"Prep Time:        {prep_time} minutes (longest item)")

    def place_order(self):
        if not self.items:
            print("Cannot place an empty order.")
            return
        subtotal, discount, total, prep_time = self.calculate_bill()
        print()
        print("===== ORDER PLACED =====")
        print(f"Items: {len(self.items)}")
        print(f"Total: Rs. {total:.0f}")
        print(f"Estimated Time: {prep_time} minutes")
        print("Your food is being prepared!")
        self.items.clear()


# ---- Menu actions ----
order = Order()


def add_item():
    print()
    print("===== ADD FOOD =====")
    print("1. Pizza  (Rs. 1200, 10% off)")
    print("2. Burger (Rs. 500, 5% off)")
    print("3. Drink  (Rs. 150)")

    choice = input("Enter choice (1-3): ").strip()
    if choice not in ("1", "2", "3"):
        print("Invalid choice.")
        return

    try:
        qty = int(input("Enter quantity: "))
        if qty <= 0:
            print("Quantity must be greater than 0.")
            return
    except ValueError:
        print("Invalid quantity.")
        return

    if choice == "1":
        order.add_food(Pizza(quantity=qty))
    elif choice == "2":
        order.add_food(Burger(quantity=qty))
    else:
        order.add_food(Drink(quantity=qty))


def remove_item():
    if not order.items:
        print("Order is empty.")
        return

    order.show_order()
    try:
        idx = int(input("Enter item number to remove: ")) - 1
    except ValueError:
        print("Invalid number.")
        return
    order.remove_food(idx)


def view_order():
    order.show_order()


def place_order():
    order.place_order()


def main():
    print("=========================================")
    print("        FOOD DELIVERY APP                ")
    print("=========================================")

    while True:
        print()
        print("===== MENU =====")
        print("1. Add Food")
        print("2. Remove Food")
        print("3. View Order")
        print("4. Place Order")
        print("5. Exit")

        choice = input("Enter choice (1-5): ").strip()

        if choice == "1":
            add_item()
        elif choice == "2":
            remove_item()
        elif choice == "3":
            view_order()
        elif choice == "4":
            place_order()
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1-5.")


main()
```
